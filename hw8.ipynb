- 作業說明：整理含有符號與空白的聚餐資料，轉換數值型別後計算服務費、折價後總額、每人分攤費用，並判斷是否超過每人 300 元預算，最後完成三組測試。

```python
# 測試一：原始資料（4 人、折價券 100 元）
event_name = "weekend dinner"
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

cleaned_event_name = event_name.strip().lower()

cleaned_meal_str = meal_amount.replace("NT$", "").replace(",", "").strip()
cleaned_meal_amount = float(cleaned_meal_str)

num_people = int(people)

service_fee = cleaned_meal_amount * service_rate
total_amount = cleaned_meal_amount + service_fee - discount
cost_per_person = total_amount / num_people

is_over_budget = cost_per_person > budget_per_person

print(f"聚餐名稱: {cleaned_event_name} ===")
print(f"原始餐費: ${cleaned_meal_amount:.2f}")
print(f"服務費: ${service_fee:.2f}")
print(f"折價券折抵: ${discount:.2f}")
print(f"應付總額: ${total_amount:.2f}")
print(f"分攤人數: {num_people} 人")
print(f"每人費用: ${cost_per_person:.2f}")
print(f"每人預算: ${budget_per_person:.2f}")
print(f"是否超過預算: {is_over_budget}")
```
